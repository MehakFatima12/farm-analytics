```python
import pandas as pd
from sqlalchemy import create_engine
import glob
import os

engine = create_engine('mysql+pymysql://root:2003%40mehak@localhost/farm_feasibility')

csv_files = glob.glob(r'F:\Mehak Fatima\Farm Analytics\Data\Clean Data Crops\*.csv')

for file in csv_files:
    table_name = os.path.splitext(os.path.basename(file))[0]  # filename = table name
    df = pd.read_csv(file)
    df.to_sql(table_name, engine, if_exists='replace', index=False)
    print(f"✅ Imported: {table_name} ({len(df)} rows)")
```
