```python
import pandas as pd
import numpy as np
```

```python
file_path = r"F:\Mehak Fatima\Farm Analytics\Data\Raw Data Crops\punjab_crop_share_vs_pakistan.csv"

raw = pd.read_csv(file_path, header=None)

raw.head(15)
```

```python
raw.info()
```

```python
years = raw.iloc[0, 1:].tolist()

years
```

```python
crops = [
    "Wheat",
    "Rice",
    "Cotton",
    "Sugarcane",
    "Maize",
    "Gram",
    "Jowar",
    "Barley",
    "Rapeseed & Mustard",
    "Tobacco",
    "Sesamum",
    "Bajra"
]

crops
```

```python
records = []

current_crop = None
current_metric = None

for i in range(1, len(raw)):
    
    label = str(raw.iloc[i, 0]).strip()
    
    # Crop
    if label in crops:
        current_crop = label
        current_metric = None
        continue
    
    # Metric
    if label.startswith("Area"):
        current_metric = "Area"
        continue
    
    if label.startswith("Production"):
        current_metric = "Production"
        continue
    
    # Geography
    if label in ["Pakistan", "Punjab"]:
        geography = label
        metric_name = current_metric
        
    elif "share" in label.lower():
        geography = "Punjab"
        metric_name = current_metric + " Share"
        
    else:
        continue
    
    # Years ke values
    for j, year in enumerate(years, start=1):
        
        value = raw.iloc[i, j]
        
        records.append({
            "Fiscal Year": year,
            "Crop": current_crop,
            "Metric": metric_name,
            "Geography": geography,
            "Value": value
        })
```

```python
df = pd.DataFrame(records)

df.head(20)
```

```python
df["Value"] = (
    df["Value"]
    .astype(str)
    .str.replace(",", "", regex=False)
    .replace("-", np.nan)
)

df["Value"] = pd.to_numeric(df["Value"], errors="coerce")
```

```python
df["Start Year"] = (
    df["Fiscal Year"]
    .str[:4]
    .astype(int)
)

df.head()
```

```python
df = df[
    [
        "Fiscal Year",
        "Start Year",
        "Crop",
        "Metric",
        "Geography",
        "Value"
    ]
]
```

```python
df = df.sort_values(
    ["Start Year", "Crop", "Metric", "Geography"]
).reset_index(drop=True)

df.head(20)
```

```python
output_path = r"F:\Mehak Fatima\Farm Analytics\Data\Clean Data Crops\punjab_crop_share_vs_pakistan_cleaned.csv"

df.to_csv(output_path, index=False)

print("Cleaned file saved successfully!")
```
