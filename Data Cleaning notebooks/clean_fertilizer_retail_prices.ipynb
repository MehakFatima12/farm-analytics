```python
import pandas as pd
```

```python
df = pd.read_excel(r"F:\Mehak Fatima\Farm Analytics\Data\Raw Data Crops\fertilizer_retail_prices.xlsx")
df.columns = ['fiscal_year', 'urea', 'can', 'ammonium_sulphate', 'nitrophosphate',
              'ssp', 'dap', 'sop']
```

```python
df = df.dropna(subset=['fiscal_year'])
df = df[~df['fiscal_year'].astype(str).str.contains('Rs per', na=False)]
```

```python
df = df.drop(columns=['can'])
df = df.replace('-', pd.NA)
```

```python
df['fiscal_year'] = df['fiscal_year'].astype(str).str.replace(' P', '', regex=False)

df.to_csv(r'F:\Mehak Fatima\Farm Analytics\Data\Clean Data Crops\fertilizer_retail_prices_clean.csv', index=False)
```
