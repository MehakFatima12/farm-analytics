```python
import pandas as pd

df = pd.read_excel(r"F:\Mehak Fatima\Farm Analytics\Data\Raw Data Crops\land_utilization_by_district.xlsx", sheet_name='pakistan', header=None)
```

```python
df = df.iloc[7:].reset_index(drop=True)
```

```python
df['year'] = df[0].where(df[0].astype(str).str.match(r'^\d{4}-\d{2}$'))
df['year'] = df['year'].ffill()
```

```python
df = df[~df[0].astype(str).str.match(r'^\d{4}-\d{2}$')]
```

```python
df.columns = ['area_name', 'geographical_area', 'reported_area', 'forest_area',
              'not_available_for_cultivation', 'culturable_waste', 'total_cultivated',
              'current_fallow', 'net_area_sown', 'sown_more_than_once', 'total_cropped',
              'agriculture_land', 'arable_land', 'uncultivable_area', 'extra_col', 'year']
```

```python
df = df.drop(columns=['extra_col'])
```

```python
df['area_name'] = df['area_name'].replace('Total', 'Pakistan')
```

```python
df.to_csv(r'F:\Mehak Fatima\Farm Analytics\Data\Clean Data Crops\land_utilization_by_province.csv', index=False)
```
