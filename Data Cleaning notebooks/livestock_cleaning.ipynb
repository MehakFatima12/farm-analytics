```python
from sqlalchemy import create_engine, text
import urllib.parse
import glob
import os

password = urllib.parse.quote_plus('2003@mehak')
engine = create_engine(f'mysql+pymysql://root:{password}@localhost/farm_feasibility')

csv_files = glob.glob(r'F:\Mehak Fatima\Farm Analytics\Data\Data Livestock\*.csv')

new_table_names = []

for file in csv_files:
    table_name = os.path.splitext(os.path.basename(file))[0]
    table_name = table_name.replace('_clean', '')
    new_table_names.append(table_name)
    
    df = pd.read_csv(file)
    df.to_sql(table_name, engine, if_exists='replace', index=False)
    print(f"✅ Imported: {table_name} ({len(df)} rows)")
```
